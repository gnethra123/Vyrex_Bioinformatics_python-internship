```python
import matplotlib.pyplot as plt
```

```python
dna ="ATGCTGGTACCTGAACGTACCGGTTAAGGCTA";
```

```python
print(dna)
```

```python

A = dna.count("A")
T = dna.count("T")
G = dna.count("G")
C = dna.count("C")
print("DNA Nucleotide Count")
print("A:",dna.count("A"))
print("T:",dna.count("T"))
print("G:",dna.count("G"))
print("C:",dna.count("C"))
```

```python

# Matplotlib graph
bases = ["A", "T", "G", "C"]
counts = [A, T, G, C]

plt.figure(figsize=(5, 5))

bars = plt.bar(bases, counts)

plt.title("DNA Nucleotide Composition", fontsize=16)
plt.xlabel("Nucleotide", fontsize=12)
plt.ylabel("Count", fontsize=12)

for bar in bars:
    height = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        height,
        str(int(height)),
        ha="center",
        va="bottom"
    )

plt.show()
```
