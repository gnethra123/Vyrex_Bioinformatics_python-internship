```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python

df = pd.read_csv("/leukemia_gene_expression.csv")

df.head()
```

```python
print(df.shape)
print(df.columns)
```

```python
mean_expression = df.drop(columns=["Diagnosis"]).mean(axis=0)

print(mean_expression.head())
```

```python
top10 = mean_expression.sort_values(ascending=False).head(10)

print("Top 10 Highly Expressed Genes:")
print(top10)
```

```python
plt.figure(figsize=(8, 5))

top10.sort_values().plot(kind="barh")

plt.title("Top 10 Expressed Genes")
plt.xlabel("Mean Expression")
plt.ylabel("Gene")

plt.tight_layout()
plt.savefig("top_genes.png")
plt.show()
```

Gene Expression Analysis

Gene expression is the process by which information from a gene is used to produce RNA or a functional protein.
High expression values indicate that a gene is more actively expressed in the samples.
The gene ranked #1 is Gene_548, with a mean expression of 2.665977. It has the highest average expression among the analyzed genes. This suggests that Gene_548 is highly expressed in the samples, but further biological analysis is needed to determine its specific role in leukemia.
